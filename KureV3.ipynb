````python
from sentence_transformers import MultiVectorEncoder

model = MultiVectorEncoder("nlpai-lab/KURE-v2")

query = "훈민정음은 언제 만들어졌나요?"
documents = [
    "세종대왕은 1443년에 훈민정음을 창제하고 1446년에 이를 반포하였다.",
    "김치는 배추나 무를 소금에 절인 뒤 고춧가루와 젓갈을 넣어 발효시킨 음식이다."
]

query_embeddings = model.encode_query(query)
document_embeddings = model.encode_document(documents)

# 토큰 수 × 128차원
print(f"질문 벡터 형태 (토큰 수, 128차원): {query_embeddings.shape}")
print(f"첫 번째 문서 벡터 형태 (토큰 수, 128차원): {document_embeddings[0].shape}")

# MaxSim 연산으로 유사도(스코어) 계산 테스트
scores = model.similarity(query_embeddings, document_embeddings)
print(f"문서별 유사도 점수: {scores}")
````
